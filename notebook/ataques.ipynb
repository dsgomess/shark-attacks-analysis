```python
sa['Fatal (S/N)'] = sa['Fatal (S/N)'].astype("string").str.strip().str.upper()


sa['Fatal (S/N)'] = sa['Fatal (S/N)'].replace({
    "Y": "Sim",
    "N": "Não",
    "SIM": "Sim",
    "NÃO": "Não",
    "NAO": "Não",
    "UNKNOWN": pd.NA,
    "F": pd.NA,
    "#VALUE!": pd.NA,
    "2017": pd.NA,
    "<NA>": pd.NA
})
```

```python
sa['Ano'] = pd.to_numeric(sa['Ano'], errors='coerce').astype('Int64')
sa.loc[(sa['Ano'] < 1500) | (sa['Ano'] > 2026), 'Ano'] = pd.NA

sa['Ano'].value_counts(dropna=False).sort_index()
```

```python
sa['Tipo de Ataque'].value_counts(dropna=False)
```

```python
sa['Tipo de Ataque'] = sa['Tipo de Ataque'].replace({
    'Unprovoked': 'Não provocado',
    'Provoked': 'Provocado',
    'Boat': 'Embarcação',
    'Boating': 'Embarcação',
    'Sea Disaster': 'Desastre marítimo',
    'Invalid': 'Inválido'
})
sa['Tipo de Ataque'].value_counts(dropna=False)
```

```python
sa['Data'] = sa['Data'].astype(str).str.replace('Reported ', '', regex=False)
sa['Data'].value_counts(dropna=False)
```

```python
sa['Data'] = sa['Data'].astype('string')

sa["Data"] = sa["Data"].str.replace("Reported ", "", regex=False)
sa["Data"] = sa["Data"].str.replace("Last incident of ", "", regex=False)
sa["Data"] = sa["Data"].str.replace("Before ", "", regex=False)
sa["Data"] = sa["Data"].str.replace("Ca. ", "", regex=False)
sa["Data"] = sa["Data"].str.strip()

sa["Data"] = pd.to_datetime(sa["Data"], errors="coerce")

sa["Data"].value_counts(dropna=False)
```

```python
sa["idade_limpa"] = sa["Idade"].astype(str).str.lower().str.strip()
sa.loc[sa["idade_limpa"].str.contains("month"), "idade_limpa"] = (
    sa["idade_limpa"].str.extract(r"(\d+)").astype(float) / 12
)
sa["idade_limpa"] = sa["idade_limpa"].str.extract(r"(\d+\.?\d*)", expand=False)
sa["Idade"] = pd.to_numeric(sa["idade_limpa"], errors="coerce")
sa.loc[(sa["Idade"] < 0) | (sa["Idade"] > 100), "Idade"] = pd.NA
sa["Idade"].value_counts(dropna=False).sort_index()




```

```python
sa['Sexo'].value_counts(dropna=False)
```

```python
sa['Sexo'] = sa['Sexo'].astype(str).str.strip().str.upper()
sa['Sexo'] = sa['Sexo'].replace({
    'M': 'Masculino',
    'F': 'Feminino',
    'N': pd.NA,
    '.': pd.NA,
    'LLI': pd.NA,
    'NAN': pd.NA
})
sa['Sexo'].value_counts(dropna=False)
```

```python
sa['Hora'].value_counts(dropna=False)
```

```python
sa['Hora'] = sa['Hora'].astype("string").str.strip().str.lower()
import re

def limpar_hora(valor):
    if pd.isna(valor):
        return pd.NA

    # pega padrão tipo 11h00, 02h30, 19:00
    match = re.search(r'(\d{1,2})[:h](\d{2})', valor)
    if match:
        h = int(match.group(1))
        m = int(match.group(2))
        if 0 <= h <= 23 and 0 <= m <= 59:
            return f"{h:02d}:{m:02d}"

    return pd.NA
sa['Hora_Limpa'] = sa['Hora'].apply(limpar_hora)
sa['Hora_Limpa'].value_counts(dropna=False)

```

```python
sa['Espécies'] = sa['Espécies'].astype("string").str.strip().str.lower()
import re

def limpar_especie(valor):
    if pd.isna(valor):
        return pd.NA

    v = valor.lower()

    if "white shark" in v or "great white" in v:
        return "White Shark"
    if "tiger shark" in v:
        return "Tiger Shark"
    if "bull shark" in v:
        return "Bull Shark"
    if "hammerhead" in v:
        return "Hammerhead Shark"
    if "whitetip" in v:
        return "Oceanic Whitetip Shark"
    if "reef shark" in v:
        return "Reef Shark"
    if "blacktip" in v:
        return "Blacktip Shark"
    if "nurse shark" in v:
        return "Nurse Shark"
    if "blue shark" in v:
        return "Blue Shark"
    if "mako shark" in v:
        return "Mako Shark"
    if "lemon shark" in v:
        return "Lemon Shark"
    if "sand tiger" in v:
        return "Sand Tiger Shark"
    if "galapagos shark" in v:
        return "Galapagos Shark"
    if "silky shark" in v:
        return "Silky Shark"

    if "shark involvement not confirmed" in v:
        return pd.NA

    if v in ["n", ".", "unknown"]:
        return pd.NA

    return "Other Shark"
sa['Especie_Limpa'] = sa['Espécies'].apply(limpar_especie)
sa['Especie_Limpa'].value_counts(dropna=False)
```

```python
sa['Nome da Vítima'] = sa['Nome da Vítima'].astype("string").str.strip()
import re

def limpar_nome(valor):
    if pd.isna(valor):
        return pd.NA

    v = valor.strip()

    lixo = ['male', 'female', 'boy', 'girl', 'man', 'woman', 'boat', 'unknown'  ]

    if v.lower() in lixo:
        return pd.NA

    if len(v) < 3:
        return pd.NA

    if not re.search(r"[a-zA-Z]", v):
        return pd.NA

    return v.title()
sa['Nome_Limpo'] = sa['Nome da Vítima'].apply(limpar_nome)
sa[['Nome da Vítima','Nome_Limpo']].sample(10)
sa['Nome_Limpo'].value_counts(dropna=False)
```
